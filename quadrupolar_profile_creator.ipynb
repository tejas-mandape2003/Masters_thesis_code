```python
import pandas as pd
```

```python
df1 = pd.read_csv("sunspots_input.dat", sep=r"\s+", header=None)
df1.columns = ['phase', 'latitude', 'tilt','radius in Mm','longitude', 'flux in Mx']
df1['flux in Mx'] = df1['flux in Mx'] * 0.25 * 1e16
df1['radius in Mm'] = df1['radius in Mm'] * 1e1 
df1.loc[df1['latitude'] < 0, 'flux in Mx'] *= -1
df1.loc[df1['latitude'] < 0, 'tilt'] *= -1
```
