```python
import numpy as np

data = np.loadtxt("sunspots_input.dat")

output = []

months = np.unique(data[:, 0].astype(int))

for m in months:
    rows = data[data[:, 0] == m]

    np.random.shuffle(rows)

    n_pairs = (len(rows) + 1) // 2

    rows = rows[:n_pairs]

    for row in rows:
        lat = abs(row[1])

        north = row.copy()
        south = row.copy()

        north[1] =  lat
        south[1] = -lat

        output.append(north)
        output.append(south)

output = np.array(output)

np.savetxt(
    "sunspots_input_2.dat",
    output,
    fmt="%.6f"
)
```
